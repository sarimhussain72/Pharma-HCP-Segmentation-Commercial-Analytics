# Pharmaceutical HCP Segmentation & Commercial Analytics

## Business Report

This notebook summarizes the business problem, analytical findings, SQL insights, and strategic recommendations derived from the pharmaceutical HCP segmentation project.

# Executive Summary

Pharmaceutical companies often rely on prescribing volume to prioritize healthcare professionals (HCPs). However, prescribing behavior alone does not explain physician mindset, adoption readiness, or promotional effectiveness.

This project combines prescription transactions, physician master data, product information, and survey responses to analyze commercial performance and HCP characteristics.

Using Python for data generation, cleaning, and visualization, along with MySQL for business analysis, the project identifies high-value physicians, strong-performing therapy areas, regional opportunities, and physician engagement patterns.

The findings support data-driven commercial decision-making and targeted physician engagement.

# Business Problem

Commercial teams need to answer questions such as:

- Which HCPs generate the highest commercial value?
- Which therapy areas contribute the most revenue?
- Which regions should receive additional field force support?
- Which products drive the majority of sales?
- How do physician characteristics influence prescribing behavior?

The objective is to transform raw operational data into actionable business insights.

# Dataset Overview

The project uses four datasets:

### HCP Master
Contains physician demographics and professional information.

### Product Master
Contains product and therapy area information.

### Prescription Transactions
Contains prescription volume, sales calls, samples, and revenue.

### Survey Data
Contains physician mindset, digital engagement, and brand awareness.

# Data Cleaning

The following data quality issues were intentionally introduced and resolved:

- Missing values
- Duplicate records

- Negative sales values
- Text inconsistencies
- Extra spaces
- Mixed letter cases

Python (Pandas) was used to clean and validate all datasets before analysis.

# Strategic Recommendations

1. Focus promotional efforts on high-performing HCPs.

2. Increase engagement in high-potential regions.

3. Allocate additional resources to strong-performing therapy areas.

4. Improve digital engagement among physicians with low adoption.

5. Personalize commercial messaging based on physician mindset.

6. Monitor prescription trends regularly for timely commercial action.

```python
hcp = pd.read_csv("../data/raw/hcp_master_generated.csv")

products = pd.read_csv("../data/raw/products_raw.csv")

prescriptions = pd.read_csv("../data/raw/prescriptions_raw.csv")

survey = pd.read_csv("../data/raw/survey_raw.csv")
```

```python
# ==========================================
# Initial Data Inspection
# ==========================================

datasets = {
    "HCP": hcp,
    "Products": products,
    "Prescriptions": prescriptions,
    "Survey": survey
}

for name, df in datasets.items():

    print("=" * 70)
    print(name)
    print("=" * 70)

    print(f"\nShape: {df.shape}")

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nDuplicate Rows:", df.duplicated().sum())

    print("\nData Types:")
    print(df.dtypes)

    print("\n")
```

```python
# ==========================================
# Create Working Copies
# ==========================================

hcp_dirty = hcp.copy()
products_dirty = products.copy()
prescriptions_dirty = prescriptions.copy()
survey_dirty = survey.copy()

print("Working copies created successfully!")
```

```python
import numpy as np

# HCP
hcp_dirty.loc[np.random.choice(hcp_dirty.index, 50, replace=False), "Age"] = np.nan
hcp_dirty.loc[np.random.choice(hcp_dirty.index, 40, replace=False), "City"] = np.nan
hcp_dirty.loc[np.random.choice(hcp_dirty.index, 30, replace=False), "Hospital_Type"] = np.nan

# Products
products_dirty.loc[np.random.choice(products_dirty.index, 2, replace=False), "Unit_Price_INR"] = np.nan

# Prescriptions
prescriptions_dirty.loc[np.random.choice(prescriptions_dirty.index, 500, replace=False), "TRx"] = np.nan
prescriptions_dirty.loc[np.random.choice(prescriptions_dirty.index, 300, replace=False), "Sales_Value"] = np.nan

# Survey
survey_dirty.loc[np.random.choice(survey_dirty.index, 20, replace=False), "Brand_Awareness"] = np.nan
survey_dirty.loc[np.random.choice(survey_dirty.index, 15, replace=False), "Digital_Engagement"] = np.nan

print("Missing values introduced successfully!")
```

```python
print("HCP")
print(hcp_dirty.isnull().sum())

print("\nProducts")
print(products_dirty.isnull().sum())

print("\nPrescriptions")
print(prescriptions_dirty.isnull().sum())

print("\nSurvey")
print(survey_dirty.isnull().sum())
```

```python
# ==========================================
# Introduce Duplicate Records
# ==========================================

# HCP - duplicate 20 rows
hcp_dirty = pd.concat(
    [hcp_dirty, hcp_dirty.sample(20, random_state=42)],
    ignore_index=True
)

# Prescriptions - duplicate 100 rows
prescriptions_dirty = pd.concat(
    [prescriptions_dirty, prescriptions_dirty.sample(100, random_state=42)],
    ignore_index=True
)

# Survey - duplicate 10 rows
survey_dirty = pd.concat(
    [survey_dirty, survey_dirty.sample(10, random_state=42)],
    ignore_index=True
)

print("Duplicate rows added successfully!")
```

```python
# ==========================================
# Invalid Ages
# ==========================================

invalid_index = np.random.choice(hcp_dirty.index, 10, replace=False)

hcp_dirty.loc[invalid_index[:5], "Age"] = 150
hcp_dirty.loc[invalid_index[5:], "Age"] = -10

print("Invalid ages introduced.")
```

```python
# ==========================================
# Text Formatting Issues
# ==========================================

# Specialty
hcp_dirty.loc[np.random.choice(hcp_dirty.index, 25), "Specialty"] = "cardiology"
hcp_dirty.loc[np.random.choice(hcp_dirty.index, 25), "Specialty"] = "CARDIOLOGY"
hcp_dirty.loc[np.random.choice(hcp_dirty.index, 25), "Specialty"] = " Cardiology "

# Therapy Area
products_dirty.loc[0, "Therapy_Area"] = "cardiovascular"
products_dirty.loc[1, "Therapy_Area"] = " CARDIOVASCULAR "

# Survey
survey_dirty.loc[np.random.choice(survey_dirty.index, 20), "Brand_Awareness"] = "high"
survey_dirty.loc[np.random.choice(survey_dirty.index, 20), "Brand_Awareness"] = " HIGH "
```

```python
# ==========================================
# Invalid Sales
# ==========================================

negative_index = np.random.choice(
    prescriptions_dirty.index,
    50,
    replace=False
)

prescriptions_dirty.loc[
    negative_index,
    "Sales_Value"
] *= -1

print("Negative sales values introduced.")
```

```python
print("HCP Shape:", hcp_dirty.shape)
print("Products Shape:", products_dirty.shape)
print("Prescriptions Shape:", prescriptions_dirty.shape)
print("Survey Shape:", survey_dirty.shape)

print("\nDuplicate Rows")
print("HCP:", hcp_dirty.duplicated().sum())
print("Products:", products_dirty.duplicated().sum())
print("Prescriptions:", prescriptions_dirty.duplicated().sum())
print("Survey:", survey_dirty.duplicated().sum())

print("\nMissing Values")
print(hcp_dirty.isnull().sum())
```

```python
# ==========================================
# Remove Duplicate Rows
# ==========================================

hcp_clean = hcp_dirty.drop_duplicates().copy()

products_clean = products_dirty.drop_duplicates().copy()

prescriptions_clean = prescriptions_dirty.drop_duplicates().copy()

survey_clean = survey_dirty.drop_duplicates().copy()

print("Duplicates removed successfully!")
```

```python
# ==========================================
# Handle Missing Values
# ==========================================

# HCP
hcp_clean["Age"] = hcp_clean["Age"].fillna(hcp_clean["Age"].median())

hcp_clean["Hospital_Type"] = hcp_clean["Hospital_Type"].fillna("Unknown")

hcp_clean["City"] = hcp_clean["City"].fillna("Unknown")

# Products
products_clean["Unit_Price_INR"] = products_clean["Unit_Price_INR"].fillna(
    products_clean["Unit_Price_INR"].median()
)

# Prescriptions
prescriptions_clean["TRx"] = prescriptions_clean["TRx"].fillna(
    prescriptions_clean["TRx"].median()
)

prescriptions_clean["Sales_Value"] = prescriptions_clean["Sales_Value"].fillna(
    prescriptions_clean["Sales_Value"].median()
)

# Survey
survey_clean["Brand_Awareness"] = survey_clean["Brand_Awareness"].fillna("Medium")

survey_clean["Digital_Engagement"] = survey_clean["Digital_Engagement"].fillna("Medium")

print("Missing values handled successfully!")
```

```python
# ==========================================
# Fix Invalid Ages
# ==========================================

median_age = hcp_clean.loc[
    (hcp_clean["Age"] >= 28) &
    (hcp_clean["Age"] <= 68),
    "Age"
].median()

hcp_clean.loc[
    (hcp_clean["Age"] < 28) |
    (hcp_clean["Age"] > 68),
    "Age"
] = median_age

print("Invalid ages corrected!")
```

```python
# ==========================================
# Standardize Text Columns
# ==========================================

def clean_text(series):
    return series.astype(str).str.strip().str.title()

hcp_clean["Specialty"] = clean_text(hcp_clean["Specialty"])
hcp_clean["Hospital_Type"] = clean_text(hcp_clean["Hospital_Type"])
hcp_clean["Practice_Type"] = clean_text(hcp_clean["Practice_Type"])
hcp_clean["City"] = clean_text(hcp_clean["City"])

products_clean["Therapy_Area"] = clean_text(products_clean["Therapy_Area"])

survey_clean["Brand_Awareness"] = clean_text(survey_clean["Brand_Awareness"])
survey_clean["Digital_Engagement"] = clean_text(survey_clean["Digital_Engagement"])

print("Text standardized!")
```

```python
# ==========================================
# Fix Negative Sales
# ==========================================

prescriptions_clean["Sales_Value"] = prescriptions_clean["Sales_Value"].abs()

print("Negative sales corrected!")
```

```python
print("========== CLEAN DATA VALIDATION ==========\n")

datasets = {
    "HCP": hcp_clean,
    "Products": products_clean,
    "Prescriptions": prescriptions_clean,
    "Survey": survey_clean
}

for name, df in datasets.items():
    print(f"{name}")
    print("-" * 40)
    print("Shape:", df.shape)
    print("Missing Values:", df.isnull().sum().sum())
    print("Duplicate Rows:", df.duplicated().sum())
    print()
```

```python
hcp_clean.to_csv("../data/cleaned/hcp_master_cleaned.csv", index=False)

products_clean.to_csv("../data/cleaned/products_cleaned.csv", index=False)

prescriptions_clean.to_csv("../data/cleaned/prescriptions_cleaned.csv", index=False)

survey_clean.to_csv("../data/cleaned/survey_cleaned.csv", index=False)

print("All cleaned datasets saved successfully!")
```

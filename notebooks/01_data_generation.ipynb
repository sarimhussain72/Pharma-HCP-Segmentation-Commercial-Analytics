# Generate HCP Master Dataset

This section generates the Healthcare Professional (HCP) master dataset.

Each HCP will have:
- Unique ID
- Name
- Gender
- Age
- Experience
- Specialty
- Hospital Type
- Practice Type
- Patient Volume
- Region
- State
- City
- Tier

```python
# ==========================================
# Generate HCP Master Dataset
# ==========================================

def generate_hcp_master(num_hcps):

    hcp_data = []

    specialty_names = list(specialty_distribution.keys())
    specialty_weights = list(specialty_distribution.values())

    for i in range(1, num_hcps + 1):

        # -------------------------
        # HCP ID
        # -------------------------
        hcp_id = f"HCP{i:05d}"

        # -------------------------
        # Name
        # -------------------------
        name = fake.name()

        # -------------------------
        # Gender
        # -------------------------
        gender = random.choice(["Male", "Female"])

        # -------------------------
        # Age
        # -------------------------
        age = random.randint(28, 68)

        # -------------------------
        # Experience
        # -------------------------
        experience = max(1, age - random.randint(24, 30))

        # -------------------------
        # Specialty
        # -------------------------
        specialty = random.choices(
            specialty_names,
            weights=specialty_weights,
            k=1
        )[0]

        # -------------------------
        # Hospital Type
        # -------------------------
        hospital = random.choice(hospital_types)

        # -------------------------
        # Practice Type
        # -------------------------
        practice = random.choice(practice_types)

        # -------------------------
        # Patient Volume
        # -------------------------
        patient_vol = random.choices(
            patient_volume,
            weights=[25,45,30],
            k=1
        )[0]

        # -------------------------
        # Location
        # -------------------------
        location = location_master.sample(1).iloc[0]

        region = location["Region"]
        state = location["State"]
        city = location["City"]
        tier = location["Tier"]

        hcp_data.append([
            hcp_id,
            name,
            gender,
            age,
            experience,
            specialty,
            hospital,
            practice,
            patient_vol,
            region,
            state,
            city,
            tier
        ])

    columns = [
        "HCP_ID",
        "HCP_Name",
        "Gender",
        "Age",
        "Experience_Years",
        "Specialty",
        "Hospital_Type",
        "Practice_Type",
        "Patient_Volume",
        "Region",
        "State",
        "City",
        "Tier"
    ]

    return pd.DataFrame(hcp_data, columns=columns)
```

# Generate Prescription Dataset

This dataset represents monthly prescription transactions made by healthcare professionals (HCPs).

Each prescription record links an HCP with a pharmaceutical product and captures key commercial metrics such as total prescriptions (TRx), new prescriptions (NRx), sales calls, samples provided, and sales value.

```python
# ==========================================
# Generate Prescription Dataset
# ==========================================

# Mapping of specialty to preferred therapy area
specialty_therapy = {
    "Cardiology": "Cardiovascular",
    "Endocrinology": "Diabetes",
    "Pulmonology": "Respiratory",
    "Oncology": "Oncology",
    "Neurology": "Neurology",
    "Nephrology": "Nephrology",
    "General Medicine": None,
    "Orthopedics": None,
    "Dermatology": None,
    "Psychiatry": None,
    "Pediatrics": None,
    "Gastroenterology": None
}

months = [
    "January","February","March","April","May","June",
    "July","August","September","October","November","December"
]

prescription_data = []

for i in range(1, NUM_PRESCRIPTIONS + 1):

    hcp = hcp_master.sample(1).iloc[0]

    specialty = hcp["Specialty"]

    preferred = specialty_therapy[specialty]

    if preferred is None:
        product = products.sample(1).iloc[0]
    else:
        product = products[
            products["Therapy_Area"] == preferred
        ].sample(1).iloc[0]

    trx = random.randint(20,250)

    nrx = random.randint(5, trx)

    sales_calls = random.randint(0,12)

    samples = random.randint(0,30)

    sales_value = trx * product["Unit_Price_INR"]

    prescription_data.append([

        f"RX{i:06d}",

        hcp["HCP_ID"],

        product["Product_ID"],

        product["Product_Name"],

        product["Therapy_Area"],

        random.choice(months),

        random.choice([2025,2026]),

        trx,

        nrx,

        sales_calls,

        samples,

        sales_value

    ])

prescriptions = pd.DataFrame(

    prescription_data,

    columns=[

        "Prescription_ID",

        "HCP_ID",

        "Product_ID",

        "Product_Name",

        "Therapy_Area",

        "Month",

        "Year",

        "TRx",

        "NRx",

        "Sales_Calls",

        "Samples_Provided",

        "Sales_Value"

    ]

)

prescriptions.head()
```

# Generate HCP Survey Dataset

This dataset represents survey responses collected from a subset of Healthcare Professionals (HCPs). It captures physician attitudes, preferences, and adoption behavior, which can later be compared with prescribing patterns.

```python
# ==========================================
# Generate Survey Dataset
# ==========================================

# Select 500 unique HCPs for the survey
survey_hcps = hcp_master.sample(NUM_SURVEYED, random_state=42)

survey_data = []

for _, hcp in survey_hcps.iterrows():

    awareness = random.choice([
        "High",
        "Medium",
        "Low"
    ])

    adoption = random.choice([
        "Early Adopter",
        "Evidence Driven",
        "Conservative",
        "Cost Conscious"
    ])

    evidence = random.choice([
        "High",
        "Medium",
        "Low"
    ])

    digital = random.choice([
        "High",
        "Medium",
        "Low"
    ])

    segment = random.choice([
        "Champions",
        "Growth Potential",
        "Maintain",
        "Low Priority"
    ])

    survey_data.append([

        hcp["HCP_ID"],

        awareness,

        adoption,

        evidence,

        digital,

        segment

    ])

survey = pd.DataFrame(

    survey_data,

    columns=[

        "HCP_ID",

        "Brand_Awareness",

        "Adoption_Mindset",

        "Evidence_Preference",

        "Digital_Engagement",

        "Segment"

    ]

)

survey.head()
```

# Query 24: Top Product in Each Therapy Area

## Business Question
Which product generates the highest revenue within each therapy area?

## SQL Query

```sql
WITH RankedProducts AS (

SELECT
    Therapy_Area,
    Product_Name,
    SUM(Sales_Value) AS Revenue,
    RANK() OVER(
        PARTITION BY Therapy_Area
        ORDER BY SUM(Sales_Value) DESC
    ) AS rnk
FROM prescriptions
GROUP BY Therapy_Area, Product_Name

)

SELECT *
FROM RankedProducts
WHERE rnk = 1;
```
